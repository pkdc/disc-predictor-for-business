```python
import pandas as pd
```

```python
emails_df = pd.read_csv('../data/extracted_emails.csv')
print(emails_df.head(5))
```

```python
emails_df.info()
```

```python
emails_df.isnull().sum()
```

```python
emails_df.dropna(how='all', inplace=True)
emails_df.isnull().sum()
```

```python
# X-bcc
print(emails_df['X-bcc'].isnull().mean() * 100, '% of X-bcc is empty')
print(emails_df['X-cc'].isnull().mean() * 100, '% of X-cc is empty')

emails_df.drop(columns=['X-bcc', 'X-cc'], inplace=True)
print('remaining cols: ', emails_df.columns)
emails_df.info()

emails_df.to_csv('../data/emails_handled_missing.csv', index=False)
```
